```python
# Subplots

# Loading necessary libraries and generating some new sample data

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# Subplots

# Defining tye types of visualizations to be plotted on each axes

fig, (ax1, ax2) = plt.subplots(ncols=2,
nrows=1,
figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
# Subplots

# Adding data to each of the subplots 
# by referencing each axes label in the someaxes list.

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
# Subplots

# Modifying figure layout

# Adding some very large x axis titles to the plot

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
```

```python
# Layouts

# Using constrained layout to prevent overlapping of figure elements.
# This ensures the subplot fit neatly in the figure area. This is done
# in two ways: tight layout to adjust subplot so tick labels, axis labels
# and titles don't overlap or leave the figure area. Constrained layout
# which is more advanced and flexible than tight layout: it works similarly
# except it also fits decorations such as legends and colorbars into the figure area.

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2', 'ax3']],
figsize=(7, 4),
layout = "constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label', fontsize=18)
someaxes["ax2"].set_xlabel('Another Label', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python
# Customizing errorbar appearance

# Use errorbar() to add error line (standard deviation in this case) 
# using yerr (which specifies the vertical error bar), and fmt to 
# remove the default line style or actual data points leaving just the error.
# Changing the color of the data points

y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color='blue')
ax.errorbar(x,y2,color='red',yerr=y2_sd, fmt='none')
```

```python
# Customizing errorbar appearance

# Use errorbar() to add error line (standard deviation in this case) 
# using yerr (which specifies the vertical error bar), and fmt to 
# remove the default line style or actual data points leaving just the error.
# Changing the color of the data points. ecolor specifies the error bar color.

y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color='blue')
ax.errorbar(x,y2,color='red',yerr=y2_sd,fmt='-', ecolor='blue')
```

```python
# Customizing errorbar appearance

# Use errorbar() to add error line (standard deviation in this case) 
# using yerr (which specifies the vertical error bar), and fmt to 
# remove the default line style or actual data points leaving just the error.
# Adding elinewidth and capsize.

y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color='blue')
ax.errorbar(x,y2,color='red',yerr=y2_sd,fmt='-', ecolor='blue',
elinewidth= 4,
capsize = 6)
```

```python
# Customizing errorbar appearance

# Use errorbar() to add error line (standard deviation in this case) 
# using yerr (which specifies the vertical error bar), and fmt to 
# remove the default line style or actual data points leaving just the error.
# Adding capthick.

y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color='blue')
ax.errorbar(x,y2,color='red',yerr=y2_sd,fmt='-', ecolor='blue',
elinewidth= 4,
capsize = 6,
capthick= 4)
```

```python
# Customizing errorbar appearance

# Use errorbar() to add error line (standard deviation in this case) 
# using yerr (which specifies the vertical error bar), and fmt to 
# remove the default line style or actual data points leaving just the error.
# Adding errorevery so we don’t see error bars for every single point.

y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color='blue')
ax.errorbar(x,y2,color='red',yerr=y2_sd,fmt='-', ecolor='blue',
elinewidth= 4,
capsize = 6,
capthick= 4, errorevery=2)
```

```python
# Customizing errorbar appearance

# Use errorbar() to add error line (standard deviation in this case) 
# using yerr (which specifies the vertical error bar), and fmt to 
# remove the default line style or actual data points leaving just the error.
# Adding errorevery and changing it to 3.

y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color='blue')
ax.errorbar(x,y2,color='red',yerr=y2_sd,fmt='-', ecolor='blue',
elinewidth= 4,
capsize = 6,
capthick= 4, errorevery=3)
```

```python
# Adding images to plots

# Loading the relevant libraries

from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
# Adding images to plots

# Getting the image from the internet

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response=requests.get(url, headers=headers)
image_file=BytesIO(response.content)
image=Image.open(image_file)
image
```

```python
# Adding image to the plot

x=np.array(['Luffy','Zoro','Nami','Usopp','Sanji'])
y1=np.array([110,180,240,99,220])
y2=np.array([170,100,90,120,50])
fig,ax=plt.subplots(figsize=(7,3))
ax.plot(x,y2,color='red')
ax_image=fig.add_axes([0.1,0.11,0.15,0.35])
ax_image.imshow(image)
ax_image.axis('off')
```
