```python
# Loading the necessary libraries and making sample data

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
# Making a line plot with both y variables on the same axes

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
```

```python
# Adding a legend - specifying a label argument to the plot

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='lower right')
fig.show()
```

```python
# Modifying legend by adding arguments to the legend object

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
frameon = True, #add frame around the legend
fontsize = 12, #change font size
ncol = 2, #specify number of columns
shadow = True )
fig.show()
```

```python
# Reposition the legend outside the plot area by specifying location on the figure

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc="upper left", bbox_to_anchor =(1, 1))
fig.show()
```

## Text and Annotations

```python
# Annotating with the text () function

# Putting the data into a scatter plot

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!")
fig.show()
```

```python
# Annotating with the text () function

# Putting the data into a scatter plot

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
```

```python
# Annotating with the text () function

# Putting text into a scatter plot

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!")
fig.show()
```

```python
# Modifying text annotation by adding arguments

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!",
ha='center', ##alignment of text
color = 'red', ##modify font colour
size = 20) ##modify font size
fig.show()
```

```python
# Linking annotation to a fixed spot on the figure using transforms

fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, "Data: (1, 5)")
```

```python
# Position the annotation with reference to the data, the 
# axes, or the figure (referencing axis/figure dimensions)

fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, "Data:(1, 5)", transform=ax.transData )
ax.text(0.5, 0.1, "Axes:(0.5, 0.1)", transform=ax.transAxes )
```

```python
# Position the annotation with reference to the data, the 
# axes, or the figure (referencing axis/figure dimensions)

fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, "Data:(1, 5)", transform=ax.transData)
ax.text(0.5, 0.1, "Axes:(0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, "Figure:(0.2, 0.2)", transform=fig.transFigure)
```

```python
# Positioning text annotations using transforms

fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, "Data: (1, 5)", transform=ax.transData )
ax.text(0.5, 0.1, "Axes: (0.5, 0.1)", transform=ax.transAxes )
ax.text(0.2, 0.2, "Figure: (0.2, 0.2)", transform=fig.transFigure )
```

```python
# Annotating with arrows and text using the annotate () function
# for instance, to focus at a specific spot.

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(facecolor='black'))
fig.show()
```

```python
# Annotating with arrows and text using the annotate () function
# for instance, to focus at a specific spot.
# Using the arrow stye as a wedge

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(arrowstyle='wedge'))
fig.show()
```

```python
# Annotating with arrows and text using the annotate () function
# for instance, to focus at a specific spot.
# Using the arrow stye as fancy

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(arrowstyle='fancy'))
fig.show()
```

```python
# Annotating with arrows and text using the annotate () function
# for instance, to focus at a specific spot.
# Changing the facecolor of the text

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(arrowstyle='wedge'), color='hotpink')
fig.show()
```

```python
# Annotating with arrows and text using the annotate () function
# for instance, to focus at a specific spot.
# Using the arrow stye as a double sided
# Changing color of the text

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(arrowstyle='<->'), color='green')
fig.show()
```

```python
# Axis Labels
# Removing tick marks or labels from the axis 
# by using locator to remove ticks and labels and formatter to remove only labels

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())
```

```python
# Axis Labels
# Limiting tick marks by defining a maximum number of ticks on the axis

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
# Axis Labels
# Setting tick marks at a particular interval (in this case, 5) using MultipleLocator

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
# Axis Labels
# Rotating axis labels and horizontal alignment using xticks () to the right

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")
plt.xticks(rotation=45, ha='right')
```

```python
# Axis Labels
# Rotating axis labels and horizontal alignment using xticks () to the left

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")
plt.xticks(rotation=45, ha='left')
```

```python
# Axis Labels
# Modifying axis title and font using xlabel () and fontdict

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")
plt.xticks(rotation=45, ha='left')
ax.set_xlabel('New X Axis')
```

```python
# Axis Labels
# Modifying axis font size

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")
plt.xticks(rotation=45, ha='left')
ax.set_xlabel('New X Axis', fontsize=30)
```

```python
# Axis Labels
# Modifying axis font family

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")
plt.xticks(rotation=45, ha='left')
font1 = {'family':'serif','color':'indigo'}
ax.set_xlabel('New X Axis', fontsize=30, fontdict=font1)
```

```python
# Axis Labels
# Modifying axis font family

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc="lower right")
plt.xticks(rotation=45, ha='left')
font1 = {'family':'serif','color':'indigo'}
#ax.set_xlabel('New X Axis', fontsize=30, fontdict=font1)
plt.xlabel ('New X Axis', fontsize=30, fontdict=font1)
```
