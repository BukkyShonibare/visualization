```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Make simple data

np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
# Make basic scatter plot

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

# Notes: 

# plt is used for the subplots.
# In the above example, 5 is the actual width, and 3 is the height.

```python
# Making a basic figure with matplotlib - modifying the earlier graph

fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
# Making a basic figure with matplotlib - making a line

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
# Class activity - modifying the existing code to produce an histogram

fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
# Creating boundaries around each bar in the histogram

fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
ax.hist(y, edgecolor='black')
```

# Plot Labels

```python
# Adding axis labels and titles:

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
# Making a dictionary with the properties for the labels

font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
# Adding fontdict arguments to the previous lineplot

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
# Moving lables with the loc argument

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

# Customizing plot appearance

```python
# Modifying data points - adding color and marker arguments.

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y,
marker='*', # NEW ADDITIONS
color = "indigo") # NEW ADDITIONS
fig.show()
```

```python
# Modifying data points - adjusting the style and width of the line connecting the points

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y,
marker='*',
color = "indigo")
linestyle = '--'    # NEW ADDITIONS
linewidth = 2      # NEW ADDITIONS
fig.show()
```

```python
# Modifying data points - Colour

fig, ax = plt.subplots(figsize=(5, 3))

# Linestyle and Linewidth are new additions

ax.plot(x, y, marker='*', color='indigo', linestyle='--', linewidth=2)
```

```python
# Modifying data points - Colour (2)

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', color='#7425b9', linestyle='--', linewidth=2)
```

```python
# Modifying data points - further customaizations e.g., marker size, marker edge color, marker face color.

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',
markersize = 12, # New Addition
linestyle = '--',
linewidth = 2,
markeredgecolor = '#fa9359',    # New Addition
markerfacecolor = '#000000')    # New Addition
```

```python
# Adding grid lines to the plot

ax.grid(axis = 'y')
```

```python
# Class activity - Adding grid lines

fig, ax = plt.subplots(figsize=(5, 3))
ax.grid(axis = 'y')
ax.plot(x,y,marker='*', 
markersize = 12,
linestyle = '--',
linewidth = 2,
markeredgecolor = '#fa9359',
markerfacecolor = '#000000')
```

```python
# Class activity - Adding grid lines

fig, ax = plt.subplots(figsize=(5, 3))
ax.grid(axis = 'y', color='blue', linewidth=2, linestyle='-.')
ax.plot(x,y,marker='*', 
markersize = 12,
linestyle = '--',
linewidth = 2,
markeredgecolor = '#fa9359',
markerfacecolor = '#000000')
```

```python
# Class Activity - Phython Graph Gallery

# width of the bars
barWidth = 0.3
 
# Choose the height of the blue bars
bars1 = [10, 9, 2]
 
# Choose the height of the cyan bars
bars2 = [10.8, 9.5, 4.5]
 
# Choose the height of the error bars (bars1)
yer1 = [0.5, 0.4, 0.5]
 
# Choose the height of the error bars (bars2)
yer2 = [1, 0.7, 1]
 
# The x position of bars
r1 = np.arange(len(bars1))
r2 = [x + barWidth for x in r1]
 
# Create blue bars
plt.bar(r1, bars1, width = barWidth, color = 'blue', edgecolor = 'black', yerr=yer1, capsize=7, label='poacee')
 
# Create cyan bars
plt.bar(r2, bars2, width = barWidth, color = 'cyan', edgecolor = 'black', yerr=yer2, capsize=7, label='sorgho')
 
# general layout
plt.xticks([r + barWidth for r in range(len(bars1))], ['cond_A', 'cond_B', 'cond_C'])
plt.ylabel('height')
plt.legend()
 
# Show graphic
plt.show()
```
